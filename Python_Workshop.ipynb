**Break statement** The break statement is used to break the loop when the required condition is satisfied.

```python
a=input("Give the number to print table")
a=int(a)
i=1
while(i<=10):
  print(f"{a}*{i}={a*i}")
  if (i==5):
    break
  i=i+1
```

```python
List_of_Books=["Python","Java","C","C++"]
book=input("enter book you want in Library")
book_no=0
while(book_no < len(List_of_Books)):
  if(book==List_of_Books[book_no]):
    print("yes, book available")
    break
  book_no=book_no+1
```

**While-else statement**

```python
List_of_Books=["Python","Java","C","C++"]
book=input("enter book you want in Library")
book_no=0
while(book_no < len(List_of_Books)):
  if(book==List_of_Books[book_no]):
    print("yes, book available")
    break
  book_no=book_no+1
else:
  print("book is not available")
```

**Nested while loop**

Print first 2-5 Tables

```python
i=2
while(i<=5):
  print(f"{i} table")
  j=1
  while(j<=10):
    print(f"{i}*{j}={i*j}")
    j=j+1
  i=i+1
```

**For Loop with Continue statement**

```python
for i in range(4):
  if(i==2):
    continue
  print(f"i=", i)
```

**Nested For Loop**

```python
for i in range(5):
  for j in range(i+1):
    print("*", end=" ")
  print(" ")
```

```python
for i in range(3):
  for j in range(3):
    if(i!=j):
      print(i, j)
  print("****")
```

```python
for i in range(3):
  for j in range(3):
    if(i!=j):
      print(i, j)
      break
  print("****")
```
